```python

import yt_dlp

url = 'https://youtu.be/TpD0SHGlH0A?si=pT0pPoM04ZL749Ap'

ydl_opts = {
    'format': 'best',  # or 'bestaudio' for audio only
    'outtmpl': '%(title)s.%(ext)s',  # output filename template
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([url])
```
